**Shallow copy and deep copy**


```python
a = [i for i in range(1,6)]
print("[a] {}".format(a))
b = a
print("[b] {}".format(b))
b.append(6)
print("[a] {}".format(a))
print("[b] {}".format(b))
```

```python
a = [i for i in range(1,6)]
print("[a] {}".format(a))
b = a[:]
print("[b] {}".format(b))
b.append(6)
print("[a] {}".format(a))
print("[b] {}".format(b))
```

```python
c = a[::-1]
print("[c] {}".format(c))
```

```python
lst1 = ['a', 'b', ['ab', 'ba']]
lst2 = lst1[:]
print("[lst1] {}".format(lst1))
print("[lst2] {}".format(lst2))
lst2[2][1] = 'd'
print("[lst1] {}".format(lst1))
print("[lst2] {}".format(lst2))
```

```python
from copy import deepcopy

lst1 = ['a', 'b', ['ab', 'ba']]
lst2 = deepcopy(lst1)
lst2[2][1] = 'd'
print("[lst1] {}".format(lst1))
print("[lst2] {}".format(lst2))
```
